```python
df = pd.read_csv("../../data_popularity/extras/overall_data/person_2020_update.csv", low_memory=False)
names = df.sort_values(by='hpi', ascending=False)['slug'].unique()
len(names)
```

```python
import requests
import random
import csv

for inx, name in enumerate(names):
    
    print("inx: ", inx, "name: ", name)
    
    for year in range(2015,2024):
        
        for lang in ['en', 'ca', 'de', 'es', 'fa', 'fi', 'fr', 'it', 'lv', 'no', 'pl', 'pt', 'ro', 'ru', 'sv', 'uk', 'vi']:
                
                csvFile = open('../../data_popularity/extras/pantheon_wikipedia_data.csv', 'a')
                csvWriter = csv.writer(csvFile)

                for month, day in zip(range(1, 13), [31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]):
                    
                    if (year in [2016,2020,2024]) and (month == 2):
                        day = 29
                    
                    url_ = "https://wikimedia.org/api/rest_v1/metrics/pageviews/per-article/%s.wikipedia/all-access/all-agents/%s/monthly/%s%.2d0100/%s%.2d%.2d00"%(lang,name,year,month,year,month,day)
                    user_agent = random.choice(user_agent_list)
                    headers = {'User-Agent': user_agent}
                    response = requests.get(url_,headers=headers)
                    soup = BeautifulSoup(response.content, 'html.parser')
                    answers = json.loads(list(soup)[0])

                    if str(response) != '<Response [404]>' and str(response) != '<Response [500]>':
                        try:
                            for x in answers['items']:
                                csvWriter.writerow([lang, name, month, year, x['views']])
                                if lang == 'en': 
                                    print(lang, name, month, year, x['views'])
                        except:
                            continue
                        
```
